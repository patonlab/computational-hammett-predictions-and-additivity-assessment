```python
# import necessary packages
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.metrics import r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# define colors used in this work
paton_hot_pink = '#d00d44'
rosewood_plum = '#9B6B8E'     
rosewood_sage = '#7BA882'     
rosewood_gold = '#D4A574'       
rosewood_slate = '#5B7FA6'     
rosewood_rose = '#C9858C'        
rosewood_charcoal = '#4A4A4A'    
rosewood_cream = '#F5EDE0' 
tol_grey = '#BBBBBB'

# define plotting parameters
sns.set(context='talk', style='ticks',
        color_codes=True, rc={'legend.frameon': False})
%matplotlib inline
import matplotlib
matplotlib.rcParams['figure.dpi'] = 300
plt.rcParams["font.family"] = 'Avenir'
plt.rcParams.update({'font.size': 24})
plt.rcParams['svg.fonttype'] = 'none'
```

```python
# read in the csv files containing the sum of the experimental and predicted Hammett parameters
disub_3_4 = 'hammett_predictions_3_4.csv'
disub_3_5 = 'hammett_predictions_3_5.csv'
trisub_3_4_5 = 'hammett_predictions_3_4_5.csv'
mono = 'hammett_predictions.csv'

df_all_3_4 = pd.read_csv(disub_3_4)
df_all_3_5 = pd.read_csv(disub_3_5)
df_all_trisub = pd.read_csv(trisub_3_4_5)
df_mono = pd.read_csv(mono)
```

```python
# plot the predicted vs. experimental Hammett parameters
x = df_all_3_4['hammett_sum'].values.reshape(-1, 1)
y = df_all_3_4['predicted_hammett'].values.reshape(-1, 1)
z = df_all_3_5['hammett_sum'].values.reshape(-1, 1)
c = df_all_3_5['predicted_hammett'].values.reshape(-1, 1)
a = df_all_trisub['hammett_sum'].values.reshape(-1,1)
b = df_all_trisub['predicted_hammett'].values.reshape(-1,1)

regr_3_4 = linear_model.LinearRegression()
regr_3_4.fit(x, y)
r_value_3_4 = r2_score(y, regr_3_4.predict(x))

mae_3_4 = mean_absolute_error(y, x)

regr_3_5 = linear_model.LinearRegression()
regr_3_5.fit(z, c)
r_value_3_5 = r2_score(c, regr_3_5.predict(z))

mae_3_5 = mean_absolute_error(c, z)

regr_3_4_5 = linear_model.LinearRegression()
regr_3_4_5.fit(a, b)
r_value_3_4_5 = r2_score(b, regr_3_4_5.predict(a))

mae_3_4_5 = mean_absolute_error(b, a)

print(f'slope 3,5: {"{:.2f}".format(regr_3_5.coef_[0][0])}')
print(f'slope 3,4: {"{:.2f}".format(regr_3_4.coef_[0][0])}')
print(f'slope 3,4,5: {"{:.2f}".format(regr_3_4_5.coef_[0][0])}')

fig, (ax1, ax2, ax3) = plt.subplots(1,3, figsize=(10, 5), sharex=True, sharey=True)
ax1.axline((0, 0), slope=1, color=tol_grey, linestyle="--")
ax2.axline((0, 0), slope=1, color=tol_grey, linestyle="--")
ax3.axline((0, 0), slope=1, color=tol_grey, linestyle="--")
ax1.scatter(z, c, color=rosewood_sage, alpha=0.30, edgecolors=rosewood_sage, s=35)
ax2.scatter(x, y, color=rosewood_plum, alpha=0.30, edgecolors=rosewood_plum, s=35)
ax3.scatter(a, b, color=rosewood_rose, alpha=0.30, edgecolors=rosewood_rose, s=35)
ax1.set_xlim(-1.5, 2.5)
ax2.set_xlim(-1.5, 2.5)
ax3.set_xlim(-1.5, 2.5)
ax1.set_xticks(np.round(np.linspace(-1, 2, 4), 2))
ax2.set_xticks(np.round(np.linspace(-1, 2, 4), 2))
ax3.set_xticks(np.round(np.linspace(-1, 2, 4), 2))
ax1.set_ylim(-1.5, 2.5)
ax2.set_ylim(-1.5, 2.5)
ax3.set_ylim(-1.5, 2.5)
ax1.set_yticks(np.round(np.linspace(-1, 2, 4), 1))
ax2.set_yticks(np.round(np.linspace(-1, 2, 4), 1))
ax3.set_yticks(np.round(np.linspace(-1, 2, 4), 1))
ax1.plot(z, regr_3_5.predict(z), color="black", linewidth=1)
ax2.plot(x, regr_3_4.predict(x), color="black", linewidth=1)
ax3.plot(a, regr_3_4_5.predict(a), color="black", linewidth=1)
ax1.annotate(f'R$^2$ = {"{:.2f}".format(r_value_3_5)}', xy=(0.045,0.90), xycoords='axes fraction', fontsize=16)
ax1.annotate(f'MAE = {"{:.2f}".format(mae_3_5)}', xy=(0.045,0.83), xycoords='axes fraction', fontsize=16)
ax1.annotate(f'n = {len(z)}', xy=(0.045,0.76), xycoords='axes fraction', fontsize=16)
ax2.annotate(f'R$^2$ = {"{:.2f}".format(r_value_3_4)}', xy=(0.045,0.90), xycoords='axes fraction', fontsize=16)
ax2.annotate(f'MAE = {"{:.2f}".format(mae_3_4)}', xy=(0.045,0.83), xycoords='axes fraction', fontsize=16)
ax2.annotate(f'n = {len(x)}', xy=(0.045,0.76), xycoords='axes fraction', fontsize=16)
ax3.annotate(f'R$^2$ = {"{:.2f}".format(r_value_3_4_5)}', xy=(0.045,0.90), xycoords='axes fraction', fontsize=16)
ax3.annotate(f'MAE = {"{:.2f}".format(mae_3_4_5)}', xy=(0.045,0.83), xycoords='axes fraction', fontsize=16)
ax3.annotate(f'n = {len(a)}', xy=(0.045,0.76), xycoords='axes fraction', fontsize=16)
fig.text(0.5, 0.01, 'Experimental Hammett Parameter', ha='center', va='center', fontsize=16)
fig.text(0.04, 0.5, 'Predicted Hammett Parameter', ha='center', va='center', rotation='vertical', fontsize=16)

# plt.savefig('polysubstituted_three_panel.svg', bbox_inches='tight', dpi=300)


plt.show()
```

```python
# 95% CI on slope for each regression
X_3_4 = sm.add_constant(x)
model_3_4 = sm.OLS(y, X_3_4).fit()
slope_3_4_ci = model_3_4.conf_int(alpha=0.05)[1]  # row 1 = slope

X_3_5 = sm.add_constant(z)
model_3_5 = sm.OLS(c, X_3_5).fit()
slope_3_5_ci = model_3_5.conf_int(alpha=0.05)[1]

X_3_4_5 = sm.add_constant(a)
model_3_4_5 = sm.OLS(b, X_3_4_5).fit()
slope_3_4_5_ci = model_3_4_5.conf_int(alpha=0.05)[1]

print(f'slope 3,5: {regr_3_5.coef_[0][0]:.2f}  (95% CI: [{slope_3_5_ci[0]:.2f}, {slope_3_5_ci[1]:.2f}])')
print(f'slope 3,4: {regr_3_4.coef_[0][0]:.2f}  (95% CI: [{slope_3_4_ci[0]:.2f}, {slope_3_4_ci[1]:.2f}])')
print(f'slope 3,4,5: {regr_3_4_5.coef_[0][0]:.2f}  (95% CI: [{slope_3_4_5_ci[0]:.2f}, {slope_3_4_5_ci[1]:.2f}])')
```

```python
# define variables for monosubstituted systems
v = df_mono['hammett_parameter'].values.reshape(-1, 1)
w = df_mono['predicted_hammett'].values.reshape(-1, 1)
```

```python
# Calculate residuals
residuals_3_5 = z - c
residuals_3_4 = x - y 
residuals_3_4_5 = a - b
residuals_mono = v - w

# Create the residuals plot
plt.figure(figsize=(10, 6))
plt.axhline(0, color=tol_grey, linestyle="--", linewidth=1)

# Plot residuals for each dataset
plt.scatter(v, residuals_mono, marker="o", s=35, color=rosewood_slate, alpha=0.2, edgecolors=rosewood_slate, label='Monosubstituted')
plt.scatter(z, residuals_3_5, marker="v", s=35, color=rosewood_sage, alpha=0.4, edgecolors=rosewood_sage, label='3,5-disubstituted')
plt.scatter(x, residuals_3_4, marker="D", s=35, color=rosewood_plum, alpha=0.4, edgecolors=rosewood_plum, label='3,4-disubstituted')
plt.scatter(a, residuals_3_4_5, marker="P", s=35, color=rosewood_rose, alpha=0.4, edgecolors=rosewood_rose, label='3,4,5-trisubstituted')

# Add labels and legend
# Add best fit lines for each dataset
plt.plot(v, np.poly1d(np.polyfit(v.flatten(), residuals_mono.flatten(), 1))(v), color=rosewood_slate, linewidth=1.5)
plt.plot(x, np.poly1d(np.polyfit(x.flatten(), residuals_3_4.flatten(), 1))(x), color=rosewood_plum, linewidth=1.5)
plt.plot(z, np.poly1d(np.polyfit(z.flatten(), residuals_3_5.flatten(), 1))(z), color=rosewood_sage, linewidth=1.5)
plt.plot(a, np.poly1d(np.polyfit(a.flatten(), residuals_3_4_5.flatten(), 1))(a), color=rosewood_rose, linewidth=1.5)
slope_residuals_mono = np.polyfit(v.flatten(), residuals_mono.flatten(), 1)[0]
slope_residuals_3_4 = np.polyfit(x.flatten(), residuals_3_4.flatten(), 1)[0]
slope_residuals_3_5 = np.polyfit(z.flatten(), residuals_3_5.flatten(), 1)[0]
slope_residuals_3_4_5 = np.polyfit(a.flatten(), residuals_3_4_5.flatten(), 1)[0]

# Print the slopes
print(f'Slope of residuals (3,4): {slope_residuals_3_4:.4f}')
print(f'Slope of residuals (3,5): {slope_residuals_3_5:.4f}')
print(f'Slope of residuals (3,4,5): {slope_residuals_3_4_5:.4f}')
print(f'Slope of residuals (monosubstituted): {slope_residuals_mono:.4f}')

# Plot residuals
plt.xlabel('Experimental Hammett Parameter (Sum)')
plt.ylabel('Residuals')
plt.legend()

plt.show()
```
