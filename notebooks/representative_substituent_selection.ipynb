```python
# import necessary packages
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.metrics import r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns

# define colors used in this work
paton_hot_pink = '#d00d44'
rosewood_plum = '#9B6B8E'     
rosewood_sage = '#7BA882'     
rosewood_gold = '#D4A574'       
rosewood_slate = '#5B7FA6'     
rosewood_rose = '#C9858C'        
rosewood_charcoal = '#4A4A4A'    
rosewood_cream = '#F5EDE0' 
tol_grey = '#BBBBBB'

# define plotting parameters
sns.set(context='talk', style='ticks',
        color_codes=True, rc={'legend.frameon': False})
%matplotlib inline
import matplotlib
matplotlib.rcParams['figure.dpi'] = 300
plt.rcParams["font.family"] = 'Avenir'
plt.rcParams.update({'font.size': 24})
plt.rcParams['svg.fonttype'] = 'none'
```

```python
# read in the stereoisomer charge data
stereoisomers = 'stereoisomer_charges.csv'

df_stereoisomers = pd.read_csv(stereoisomers)
```

```python
# plot the CM5 charges of E-specified stereoisomers vs. E/Z stereoisomers 
x = df_stereoisomers['cm5_charge_mixed']
y = df_stereoisomers['cm5_charge_specified']

regr = linear_model.LinearRegression()
regr.fit(x.values.reshape(-1, 1), y.values) 
mae = mean_absolute_error(y, x)
print(f"Mean Absolute Error in Charge: {mae:.4f}")

MAE_pred = 34.53 * mae
print(f"Mean Absolute Error in predicted Hammett sigma values: {MAE_pred:.3f}")

plt.axline((0, 0), slope=1, color=tol_grey, linestyle='--', zorder=1)
plt.scatter(x, y, color=rosewood_sage, s=80, alpha=0.3)
plt.xlabel('CM5 Charge (E and Z isomers)')
plt.ylabel('CM5 Charge (E-specified isomers)')
plt.xlim(-0.36, -0.31)
plt.ylim(-0.36, -0.31)
```

```python
# read in the stereoisomer charge data
conformers = 'conformer_charges.csv'

df_conformers = pd.read_csv(conformers)
```

```python
# plot the CM5 charges of the lowest energy conformer vs. the Boltzmann weighted conformer ensemble
x = df_conformers['cm5_charge_boltzmann_weighted']
y = df_conformers['cm5_charge_lowest_energy']

regr = linear_model.LinearRegression()
regr.fit(x.values.reshape(-1, 1), y.values)  
mae = mean_absolute_error(y, x)
print(f"Mean Absolute Error in Charge: {mae:.4f}")

MAE_pred = 34.53 * mae
print(f"Mean Absolute Error in predicted Hammett sigma values: {MAE_pred:.3f}")

plt.axline((0, 0), slope=1, color=tol_grey, linestyle='--', zorder=1)
plt.scatter(x, y, color=rosewood_gold, s=30, alpha=0.3)
plt.xlabel('CM5 Charge (Boltzmann Weighted)')
plt.ylabel('CM5 Charge (Lowest Energy)')
plt.xlim(-0.36, -0.31)
plt.ylim(-0.36, -0.31)
```
