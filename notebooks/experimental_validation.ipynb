```python
# import necessary packages
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.metrics import r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns

# define colors used in this work
paton_hot_pink = '#d00d44'
rosewood_plum = '#9B6B8E'     
rosewood_sage = '#7BA882'     
rosewood_gold = '#D4A574'       
rosewood_slate = '#5B7FA6'     
rosewood_rose = '#C9858C'        
rosewood_charcoal = '#4A4A4A'    
rosewood_cream = '#F5EDE0' 
tol_grey = '#BBBBBB'

# define plotting parameters
sns.set(context='talk', style='ticks',
        color_codes=True, rc={'legend.frameon': False})
%matplotlib inline
import matplotlib
matplotlib.rcParams['figure.dpi'] = 300
plt.rcParams["font.family"] = 'Avenir'
plt.rcParams.update({'font.size': 24})
plt.rcParams['svg.fonttype'] = 'none'
```

Experimental values were taken from Hansch et al., J. Med. Chem. 1973, Table II. SMILES for di- and tri-substituted benzonitriles corresponding to structures with experimentally measured Hammett constants for the polysubstituted system. Experimentally measured polysubstituted constants correspond to the column 'observed_hammett'.

```python
# read in the data
exp_3_4 = 'experimental_values_3_4.csv'
exp_3_5 = 'experimental_values_3_5.csv'
exp_3_4_5 = 'experimental_values_3_4_5.csv'
```

```python
df_3_4 = pd.read_csv(exp_3_4)
df_3_5 = pd.read_csv(exp_3_5)
df_3_4_5 = pd.read_csv(exp_3_4_5)
```

```python
# drop the cm5 charges and hammett predictions, as we will predict these in this notebook
df_3_4.drop(columns=['cm5_charge', 'predicted_hammett'], inplace=True)
df_3_5.drop(columns=['cm5_charge', 'predicted_hammett'], inplace=True)
df_3_4_5.drop(columns=['cm5_charge', 'predicted_hammett'], inplace=True)
```

```python
# plot the correlation between the experimentally observed Hammett 3_5meters and the sum of the monosubstituted parameters
x = df_3_4['hammett_sum'].values.reshape(-1, 1)
y = df_3_4['observed_hammett'].values.reshape(-1, 1)
z = df_3_5['hammett_sum'].values.reshape(-1, 1)
c = df_3_5['observed_hammett'].values.reshape(-1, 1)
a = df_3_4_5['hammett_sum'].values.reshape(-1,1)
b = df_3_4_5['observed_hammett'].values.reshape(-1,1)

regr_3_4 = linear_model.LinearRegression()
regr_3_4.fit(x, y)
r_value_3_4 = r2_score(y, regr_3_4.predict(x))

mae_3_4 = mean_absolute_error(y, x)

regr_3_5 = linear_model.LinearRegression()
regr_3_5.fit(z, c)
r_value_3_5 = r2_score(c, regr_3_5.predict(z))

mae_3_5 = mean_absolute_error(c, z)

regr_3_4_5 = linear_model.LinearRegression()
regr_3_4_5.fit(a, b)
r_value_3_4_5 = r2_score(b, regr_3_4_5.predict(a))

mae_3_4_5 = mean_absolute_error(b, a)

print(f"slope 3,4: {regr_3_4.coef_[0][0]}")
print(f"slope 3,5: {regr_3_5.coef_[0][0]}")
print(f"slope trisub: {regr_3_4_5.coef_[0][0]}")

fig, (ax1, ax2, ax3) = plt.subplots(1,3, figsize=(10, 5), sharex=True, sharey=True)
ax1.axline((0, 0), slope=1, color=tol_grey, linestyle="--")
ax2.axline((0, 0), slope=1, color=tol_grey, linestyle="--")
ax3.axline((0, 0), slope=1, color=tol_grey, linestyle="--")
ax1.scatter(x, y, color=rosewood_plum, alpha=0.30, edgecolors=rosewood_plum, s=35)
ax2.scatter(z, c, color=rosewood_sage, alpha=0.30, edgecolors=rosewood_sage, s=35)
ax3.scatter(a, b, color=rosewood_rose, alpha=0.30, edgecolors=rosewood_rose, s=35)
ax1.set_xlim(-1.5, 2.5)
ax2.set_xlim(-1.5, 2.5)
ax3.set_xlim(-1.5, 2.5)
ax1.set_xticks(np.round(np.linspace(-1, 2, 4), 2))
ax2.set_xticks(np.round(np.linspace(-1, 2, 4), 2))
ax3.set_xticks(np.round(np.linspace(-1, 2, 4), 2))
ax1.set_ylim(-1.5, 2.5)
ax2.set_ylim(-1.5, 2.5)
ax3.set_ylim(-1.5, 2.5)
ax1.set_yticks(np.round(np.linspace(-1, 2, 4), 1))
ax2.set_yticks(np.round(np.linspace(-1, 2, 4), 1))
ax3.set_yticks(np.round(np.linspace(-1, 2, 4), 1))
ax1.plot(x, regr_3_4.predict(x), color="black", linewidth=1)
ax2.plot(z, regr_3_5.predict(z), color="black", linewidth=1)
ax3.plot(a, regr_3_4_5.predict(a), color="black", linewidth=1)
ax1.annotate(f'R$^2$ = {"{:.2f}".format(r_value_3_4)}', xy=(0.045,0.90), xycoords='axes fraction', fontsize=16)
ax1.annotate(f'MAE = {"{:.2f}".format(mae_3_4)}', xy=(0.045,0.83), xycoords='axes fraction', fontsize=16)
ax1.annotate(f'n = {len(x)}', xy=(0.045,0.76), xycoords='axes fraction', fontsize=16)
# ax1.annotate('3,4-disubstituted', xy=(0.25,0.1), xycoords='axes fraction', fontsize=14)
ax2.annotate(f'R$^2$ = {"{:.2f}".format(r_value_3_5)}', xy=(0.045,0.90), xycoords='axes fraction', fontsize=16)
ax2.annotate(f'MAE = {"{:.2f}".format(mae_3_5)}', xy=(0.045,0.83), xycoords='axes fraction', fontsize=16)
ax2.annotate(f'n = {len(z)}', xy=(0.045,0.76), xycoords='axes fraction', fontsize=16)
# ax2.annotate('3,5-disubstituted', xy=(0.25,0.1), xycoords='axes fraction', fontsize=14)
ax3.annotate(f'R$^2$ = {"{:.2f}".format(r_value_3_4_5)}', xy=(0.045,0.90), xycoords='axes fraction', fontsize=16)
ax3.annotate(f'MAE = {"{:.2f}".format(mae_3_4_5)}', xy=(0.045,0.83), xycoords='axes fraction', fontsize=16)
ax3.annotate(f'n = {len(a)}', xy=(0.045,0.76), xycoords='axes fraction', fontsize=16)
# ax3.annotate('3,4,5-trisubstituted', xy=(0.25,0.1), xycoords='axes fraction', fontsize=14)
fig.text(0.5, 0.01, 'Hammett Sum', ha='center', va='center', fontsize=16)
fig.text(0.04, 0.5, 'Observed Hammett Parameter', ha='center', va='center', rotation='vertical', fontsize=16)

plt.show()
```
